```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
train_data = r'data\train.csv'
```

```python
train_df = pd.read_csv(train_data, encoding='utf8')
```

```python
fig, ax = plt.subplots(figsize=(8,5))

max_price = train_df['SalePrice'].max()
mean_price = train_df['SalePrice'].mean()

ax.hist(train_df['SalePrice'],bins=50, color='green', histtype='step')
ax.plot([mean_price, mean_price], [0,200],'--', scaley=True, linewidth=2, color='brown')
# ax.set_xticks(np.arange(0,max_price, 50000))
ax.tick_params(axis='x',labelsize=7, pad=10)
ax.set_title('Distribution of House prices')
plt.show()
```

### Applying log on target

```python
train_df['SalePrice'] = np.log1p(train_df['SalePrice'])
train_df['SalePrice']
```

```python
df = train_df.copy()

int_cols = df.dtypes[df.dtypes == 'int64'].index.to_list()[1:]
int_cols_last_16 = int_cols[16:]
int_cols[:5]
```

```python
fig, ax = plt.subplots(nrows=4, ncols=4, figsize=(15,15))
z = 0
for x in range(0, 4):
    for y in range(0, 4):
        ex_ind = int_cols[x+y+z]
        ax[x,y].hist(df[ex_ind],density=False,bins=30, color='purple', histtype='step')
        ax[x,y].set_title(ex_ind)
    z += 3
```

```python
fig, ax = plt.subplots(nrows=4, ncols=4, figsize=(15,15))
z = 0
for x in range(0, 4):
    for y in range(0, 4):
        ex_ind = int_cols_last_16[x+y+z]
        ax[x,y].hist(df[ex_ind],density=True,bins=30, color='purple', histtype='step')
        ax[x,y].set_title(ex_ind)
    z += 3
```
