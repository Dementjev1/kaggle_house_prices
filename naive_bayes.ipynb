```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
train_data = r'data\train.csv'

df = pd.read_csv(train_data, encoding='utf8')
df = df.fillna(0)
```

```python
y = df['SalePrice']
X = df.drop(['SalePrice'],axis=1)
```

```python
str_cols = X.dtypes[X.dtypes == 'str'].index
obje_cols = X.dtypes[X.dtypes == 'object'].index
X = X.drop(list(str_cols)+list(obje_cols), axis=1)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
model = SGDRegressor(loss='squared_error')

y_pred = model.fit(X_train, y_train).predict(X_test)
print(mean_absolute_error(y_test,y_pred)) 

```
